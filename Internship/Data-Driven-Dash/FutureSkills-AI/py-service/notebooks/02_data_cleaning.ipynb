```python
import pandas as pd

df = pd.read_csv("D:/Internship/Data-Driven-Dash/FutureSkills-AI/py-service/data/ai_job_market.csv")

df.head()
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
skill_columns = [
    "skills_python",
    "skills_sql",
    "skills_ml",
    "skills_deep_learning",
    "skills_cloud"
]

df["total_skills"] = df[skill_columns].sum(axis=1)
```

```python
df["posting_date"] = (
    df["job_posting_month"].astype(str)
    + "-"
    + df["job_posting_year"].astype(str)
)
```

```python
df.to_csv(
    "../data/processed/clean_jobs.csv",
    index=False
)
```
